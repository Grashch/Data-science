```python
from hyperopt import fmin, tpe, hp, Trials
import numpy as np

# Функция, которую оптимизируем (имитация обучения модели)
def objective(params):
    # params — словарь с параметрами, например: {'lr': 0.01, 'depth': 5}
    lr = params['lr']
    depth = params['depth']

    # Имитируем качество модели: лучше, когда lr ≈ 0.05 и depth ≈ 7
    score = -(lr - 0.05)**2 - (depth - 7)**2 + np.random.normal(0, 0.1)
    return score  # Чем меньше score, тем лучше (минимизируем)


# Определяем пространство поиска параметров
space = {
    'lr': hp.loguniform('lr', np.log(0.001), np.log(0.1)),  # 0.001..0.1 (логарифм.)
    'depth': hp.quniform('depth', 3, 10, 1),                # 3, 4, ..., 10
}

np.random.seed(42)

# Запускаем оптимизацию
trials = Trials()
best = fmin(
    fn=objective,           # наша функция качества
    space=space,            # пространство параметров
    algo=tpe.suggest,       # алгоритм TPE
    max_evals=50,          # число попыток
    trials=trials           # сохраняем историю
    )

print("Лучшие параметры:", best)
```
